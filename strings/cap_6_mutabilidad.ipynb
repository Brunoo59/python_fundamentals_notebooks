## OBJETOS MUTABLES

### Listas

```python
L = [2, 3, 4]
M = L
L[0] = 24     # Cambia el primer elemento de L 

print("L = ", L)
print("M = ", M)
```

Copia de objetos en vez de referencia:

```python
L = [2, 3, 4]
M = L[:]      # Crea una copia de la lista L
L[0] = 24     # Cambia el primer elemento de L
print("L = ", L)
print("M = ", M)   # M = [2, 3, 4] porque M es una copia de L y no cambia cuando L cambia
```

### Shared References and Equality

· Operador `==` igualdad de valores.  
· Operador `is` identidad de objetos.


```python
L = [1, 2, 3]
M = L

L == M      # True porque L y M tienen los mismos elementos
L is M      # True porque L y M son el mismo objeto en memoria
```

```python
L = [1, 2, 3]
M = [1, 2, 3]

print(
    "L == M",
    L == M,  # True porque L y M tienen los mismos elementos
)

print(
    "L is M",
    L is M,  # False porque L y M son objetos diferentes en memoria
)
```

### CACHE
Because small integers and strings are cached and reused, though, is tells us they reference the same single object.

```python
X = 42
Y = 42

X == Y  # True porque X y Y tienen el mismo valor
X is Y  # True porque X y Y son el mismo objeto en memoria (los enteros pequeños son internados en Python)
```

Averiguar el numero de refencias a un objeto:

```python
import sys
sys.getrefcount(1)  # numero de referencias al objeto entero 1 en la memoria
```
